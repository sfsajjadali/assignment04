### Q5
`numbers.txt` file ko read karein aur tamam numbers ka sum (total) calculate karke print karein.

```python
# Q5 solution code here
total_sum = 0

with open('numbers.txt', 'r') as f:
    for line in f:
        words = line.split()
        for word in words:
            if word.isdigit():  
                total_sum += int(word)

print(f"Total Sum: {total_sum}")
```

### Q14 
Ek program banayein jisme `finally` block ho, jo humesha execute ho chahe error aaye ya na aaye.

```python
# Q14 solution code here
def check_number_division(num1, num2):
    try:
        print(f" Output for: {num1} / {num2}")
        result = num1 / num2
        print(f"Result: {result}")
    except ZeroDivisionError:
        print("Error Catch Ho Gaya: Zero (0) se divide nahi kiya ja sakta!")
    finally:
        print("Finally Block: Operations complete ho gaye hain.")
```

```python
check_number_division(10, 2)
```

```python
check_number_division(10, 0)
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
# Q18 solution code heredef divide_numbers(a, b):
def divide_numbers(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Zero (0) se divide nahi kiya ja sakta!")
    else:
        print(f"Success! Division ka result hai: {result}")

print("~~~ Test 1 ~~~")
divide_numbers(10, 2)

print("\n--- Test 2 ---")
divide_numbers(10, 0)
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
def validate_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    else:
        print(f"Access Granted! Aap ki umar {age} saal hai.")
try:
    validate_age(20)
except ValueError as e:
    print(f"Error caught: {e}")
try:
    validate_age(15)
except ValueError as e:
    print(f"Error caught: {e}")
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
file = "sample_data.txt"
```

```python
# Q20 solution code here
file = "sample_data.txt"
file = None
try:
    with open("sample_data.txt", "r", encoding="utf-8") as file:
        content = file.read()
        print(content)
        
except FileNotFoundError:
    print("Error: 'sample_data.txt' file exist nahi karti!")

except Exception as e:
    print(f"Koi aur unexpected error aaya: {e}")

finally:
    if file is not None and not file.closed:
        file.close()
        print("Finally Block: File ko safely close kar diya gaya hai.")
    else:
        print("Finally Block: File open nahi ho saki thi, isliye closing ki zaroorat nahi pari.")
```
